```python
import calendar
import datetime as dt
from datetime import datetime, timedelta, timezone

MSK_TIMEZONE_NAME: str = "Moscow TZ (GMT+3)"
MSK_TIMEZONE: timezone = timezone(timedelta(hours=3), name=MSK_TIMEZONE_NAME)

# - params
current_timezone: timezone = MSK_TIMEZONE
delta_years: int = -8
delta_months: int = -7
delta_days: float = -7
delta_hours: float = -17
delta_minutes: float = 0
delta_seconds: float = 0

# current timestamp
current: datetime = datetime.now(current_timezone)
year: int = current.year  # get year
month: int = current.month  # get month
day: int = current.day  # get day
hour: int = current.hour  # get hour
minute: int = current.minute  # get minute
second: int = current.second  # get second 

# adjust YEAR according to delta
if delta_years != 0:
    year += delta_years

# adjust MONTH + YEAR (if necessary)
if delta_months != 0:
    month += delta_months
    if abs(month) > 12:  # adding year(s) (roll over several months)
        year += month // 12
        month += month % 12
    elif abs(month) < 1:  # substracting year (month = 0 -> previous year, december)
        year -= 1
        month = 12
    else:  # 1 <= abs(month) <= 12. but may be negative
        if month < 0:
            year -= 1
            month = 12 + month
        
result: datetime = datetime(year, month, day, hour, minute, second) + \
    timedelta(days=delta_days, hours=delta_hours, minutes=delta_minutes, seconds=delta_seconds)

print(result)
```
