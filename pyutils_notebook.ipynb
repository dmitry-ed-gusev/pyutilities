```python
from collections import defaultdict

class KeyAwareDefaultDict(defaultdict):
    def __missing__(self, key):
        # Вызываем фабрику с ключом и сохраняем результат
        result = self.default_factory(key)
        print(key, "---", result)
        self[key] = result
        return result

def dynamic_factory(key):
    """Return different default types based on the key prefix."""

    print("***", key)

    if not key:
        return None
    
    if key.startswith('count_'):
        return 0
    elif key.startswith('text_'):
        return ''
    elif key.startswith('items_'):
        return []
    else:
        return None  # fallback for unknown keys

# Создаём экземпляр нашего класса
# dynamic_dict = KeyAwareDefaultDict(dynamic_factory)

# dynamic_dict = defaultdict(lambda: None)
# dynamic_dict.default_factory = dynamic_factory

dynamic_dict = defaultdict(dynamic_factory(None))

# Тестируем разные типы ключей — синтаксис как у обычного defaultdict
dynamic_dict['count_users'] += 5
dynamic_dict['text_welcome'] += 'Welcome!'
dynamic_dict['items_shopping'].append('milk')
dynamic_dict['items_shopping'].append('bread')

# Выводим результат для проверки
print("count_users:", dynamic_dict['count_users'])  # 5
print("text_welcome:", dynamic_dict['text_welcome'])  # 'Welcome!'
print("items_shopping:", dynamic_dict['items_shopping'])  # ['milk', 'bread']
```
